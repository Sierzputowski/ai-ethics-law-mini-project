# Odtworzenie wyników

Ten notebook uruchamia główny skrypt analizy HITL i sprawdza, czy powstały wszystkie wymagane artefakty w katalogu `wyniki/`.

```python
from pathlib import Path
import subprocess
import sys

ROOT = Path.cwd().resolve().parent if Path.cwd().name == "notebooks" else Path.cwd().resolve()
ROOT
```

```python
wynik = subprocess.run(
    [sys.executable, str(ROOT / "src" / "main.py")],
    cwd=ROOT,
    text=True,
    capture_output=True,
    check=True,
)
print(wynik.stdout)
```

```python
wymagane = [
    "scenariusze.csv",
    "wyniki_polityk_hitl.csv",
    "podsumowanie_polityk.csv",
    "rekomendacje_polityk.csv",
    "analiza_obciazenia_operatora.csv",
    "macierz_zgodnosci_ai_act.csv",
    "raport_hitl.md",
    "ryzyko_resztkowe_polityki.svg",
    "kompromis_ryzyko_czas.svg",
    "heatmapa_wysokiego_ryzyka.svg",
]

for nazwa in wymagane:
    sciezka = ROOT / "wyniki" / nazwa
    print(f"{nazwa}: {'OK' if sciezka.exists() else 'BRAK'}")
```

Jeżeli wszystkie pliki mają status `OK`, wyniki projektu zostały odtworzone. Interpretacja znajduje się w notebooku `analyse.ipynb`.
